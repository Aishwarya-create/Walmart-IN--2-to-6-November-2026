```python
# Traditional Search vs Semantic Search:

# Q. "Can I take leave if my child is sick?"

# Assume the company's HR policy document contains this sentence:
#     "Employees are eligible for up to 3 days of dependent care leave annually to attend to immediate family members requiring medical assistance."

# 1. Traditional Search (Keyword-Based)
# Searches for matching words such as "child", "sick", and "leave".

# EXAMPLE OUTPUT
# "No exact matching policy found for 'child sick leave'."

# Why? The document uses "dependent care", "family members" and "medical assistance" instead of the user's words. A basic keyword search may miss the relevant policy.


# 2. Semantic Search (Meaning-Based)
# Understands that "sick child" relates to "dependent care" and "medical assistance".

# EXAMPLE OUTPUT
#     "Relevant policy found: Employees may take up to 3 days of dependent care leave annually for immediate family members requiring medical assistance."

# Why? Semantic search uses embeddings to identify conceptually similar text, even when the exact words differ.
```

| Feature             | Traditional Search            | Semantic Search                                          |
| ------------------- | ----------------------------- | -------------------------------------------------------- |
| Searches based on   | Keywords                      | Meaning                                                  |
| Exact words needed? | Often important               | Not necessarily                                          |
| Synonyms            | Limited without expansion     | Usually recognized                                       |
| Technology          | Inverted index, BM25          | Embeddings, vector similarity                            |
| Best for            | IDs, error codes, exact terms | Natural-language questions                               |
| Typical weakness    | Misses different wording      | May retrieve conceptually related but incorrect passages |

```python
# Real-Life Example: Searching on Amazon

# Prompt: "I want shoes for running."
```

| Traditional Search                        | Semantic Search                                                |
| ----------------------------------------- | -------------------------------------------------------------- |
| Looks for the words "shoes" and "running" | Understands you need footwear suitable for running             |
| Finds products titled "Running Shoes"     | Finds "Running Shoes", "Jogging Sneakers", "Athletic Footwear" |
| May miss products without exact keywords  | Can find products even with different wording                  |

```python
# - Traditional Search: Matches the words you type.
# - Semantic Search: Matches the meaning of what you ask.

# Remember: Traditional search is like finding a word in a textbook. Semantic search is like asking a knowledgeable teacher to find the relevant topic.
```
